```python
def create_churn_labels(df):

    df = df.sort_values(
        ["Loyalty Number", "Date"]
    ).copy()

    churn_labels = []

    for customer_id, group in df.groupby("Loyalty Number"):

        engagement = group["Engaged"].values

        labels = []

        for i in range(len(group)):

            future_window = engagement[i+1:i+7]

            if len(future_window) < 6:
                labels.append(np.nan)

            elif future_window.sum() == 0:
                labels.append(1)

            else:
                labels.append(0)

        churn_labels.extend(labels)

    df["Churn"] = churn_labels

    return df
```

```python
def retention_action(row):

    segment = row["Segment_Name"]
    risk = row["Risk_Level"]

    if segment == "Champions" and risk in ["High","Critical"]:
        return "VIP Outreach + Tier Benefits"

    elif segment == "Loyal Regulars" and risk in ["High","Critical"]:
        return "Bonus Points Campaign"

    elif segment == "Dormant Members":
        return "Win-Back Offer"

    elif segment == "Established Loyalists" and risk in ["Medium","High"]:
        return "Anniversary Reward"

    else:
        return "No Immediate Action"
```

```python
customer_risk["Recommended_Action"] = (
    customer_risk.apply(
        retention_action,
        axis=1
    )
)
```
